## 3. Analysis and Visualizations
تحليل الأسئلة الثلاثة المطلوبة وعرض نتيجة كل سؤال مع رسم بياني بسيط.

### Question 1: Which product category has the highest number of orders?
**السؤال الأول: أنهي فئة منتجات هي الأكثر مبيعًا من حيث عدد الطلبات؟**

We count the number of orders in each category and identify the category with the highest count.

```python
category_counts = df["Category"].value_counts()
top_category = category_counts.idxmax()
top_category_count = category_counts.max()

print(f"Most ordered category: {top_category}")
print(f"Number of orders: {top_category_count}")

category_counts.plot(kind="bar", title="Number of Orders by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Orders")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

**Result:** The category with the highest number of orders is **Beauty**, with **11 orders**.

**النتيجة:** فئة **Beauty (الجمال)** هي الأكثر طلبًا، بإجمالي **11 طلبًا**.

### Question 2: What is the average order value?
**السؤال الثاني: ما متوسط قيمة الطلب بشكل عام؟**

We calculate the arithmetic mean of the `Order Amount (EGP)` column.

```python
average_order_value = df["Order Amount (EGP)"].mean()

print(f"Average order value: {average_order_value:,.2f} EGP")

plt.figure()
plt.bar(["Average Order Value"], [average_order_value])
plt.title("Average Order Value")
plt.ylabel("Amount (EGP)")
plt.text(0, average_order_value, f"{average_order_value:,.2f} EGP",
         ha="center", va="bottom")
plt.tight_layout()
plt.show()
```

**Result:** The average order value is **EGP 646.31** (approximately).

**النتيجة:** متوسط قيمة الطلب الواحد حوالي **646.31 جنيه مصري**.

### Question 3: Which payment method is used most frequently?
**السؤال الثالث: ما طريقة الدفع الأكثر استخدامًا من العملاء؟**

We count the orders for each payment method and identify the most frequently used method.

```python
payment_counts = df["Payment Method"].value_counts()
most_used_payment = payment_counts.idxmax()
most_used_payment_count = payment_counts.max()

print(f"Most used payment method: {most_used_payment}")
print(f"Number of orders: {most_used_payment_count}")

payment_counts.plot(kind="bar", title="Number of Orders by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()
```

**Result:** **Wallet** is the most frequently used payment method, with **14 orders**.

**النتيجة:** طريقة الدفع الأكثر استخدامًا هي **Wallet (المحفظة الإلكترونية)**، بإجمالي **14 طلبًا**.
